```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('/content/customer_shopping_behavior.csv')
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(' ','_')
df.info()
```

```python
labels =['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels= labels)
df.info()
df[['age','age_group']].head()
```

```python
frequency_mapping = {
'Fortnightly': 14,
'Weekly': 7,
'Monthly': 38,
'Quarterly': 90,
'Bi-Weekly': 14,
'Annually': 365,
'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head()
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df.drop(columns=['promo_code_used'], inplace=True)
```

```python
df.to_csv("cleaned_customer_behavior.csv", index=False)
```
